```python
import requests
import time

page = 1
all_anime = []

while True:
    url = f"https://api.jikan.moe/v4/anime?page={page}"
    response = requests.get(url)

    if response.status_code != 200:
        print("Error:", response.status_code)
        break

    data = response.json()

    if not data["data"]:
        break  # kalau sudah habis

    all_anime.extend(data["data"])
    print(f"Page {page} berhasil diambil, total: {len(all_anime)} anime")

    page += 1
    time.sleep(3)  # hindari rate limit (disarankan min 3 detik/request)

print("Total data terkumpul:", len(all_anime))
```

```python
import csv

with open("mal_anime_dataset.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["mal_id", "title", "type", "episodes", "score", "status", "year"])

    for anime in all_anime:
        writer.writerow([
            anime["mal_id"],
            anime["title"],
            anime.get("type"),
            anime.get("episodes"),
            anime.get("score"),
            anime.get("status"),
            anime.get("year")
        ])
```
